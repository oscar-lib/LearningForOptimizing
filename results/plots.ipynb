```python
import orjson
from dataclasses import dataclass


@dataclass
class SolutionPoint:
    t: float
    v: float


def parse_solution_over_time(sol_over_time: str) -> list[SolutionPoint]:
    """
    Parse the solution over time that is almost a JSON (commas have been replaced by hyphens for CSV encoding)
    """
    sol_over_time = sol_over_time.replace("-", ",")
    sol_over_time = orjson.loads(sol_over_time)
    return [SolutionPoint(**point) for point in sol_over_time] #type: ignore
```
